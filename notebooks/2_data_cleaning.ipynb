## 2. Data Cleaning 

```python
import pandas as pd
import numpy as np
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
PROJECT_ROOT = os.getenv('PROJECT_ROOT')
df = pd.read_csv(f"{PROJECT_ROOT}/data/raw/student_performance_data.csv")
df.shape
```

```python
df.head()
```

```python
df.info()
```

```python
df['Age'].value_counts()
```

```python
df['Age'] = df['Age'].astype('uint8')
```

```python
df['Class'].value_counts()
```

```python
df['Class'] = df['Class'].astype('category')
df.sample()
```

```python
df['Attendance_Percentage'].unique()
```

```python
df['Attendance_Percentage'] = df['Attendance_Percentage'].astype('uint8')
```

```python
df = pd.get_dummies(df, columns = ['Parental_Education'], prefix = 'Parent_Education')
df.head()
```

```python
internet_access_map = {"Yes" : 1, "No" : 0}
extra_activities_map = {"Yes" : 1, "No" : 0}
df['Internet_Access'] = df["Internet_Access"].map(internet_access_map)
df['Extracurricular_Activities'] = df["Extracurricular_Activities"].map(extra_activities_map)
df.head()
```

```python
print(df['Final_Percentage'].min())
print(df['Final_Percentage'].max())
```

```python
# Is Final_Percentage the mean of the three scores?
(df[['Math_Score', 'Science_Score', 'English_Score']].mean(axis=1) - df['Final_Percentage']).abs().max()

# Is Pass_Fail just a threshold on Final_Percentage?
df.groupby('Pass_Fail')['Final_Percentage'].agg(['min', 'max'])
```

```python
numeric_cols = df.select_dtypes(include='number').columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - (1.5 * IQR)
    upper_bound = Q3 + (1.5 * IQR)

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    outlier_count = len(outliers)
    outlier_percentage = (outlier_count / len(df)) * 100

    print(f"{col}:")
    print(f"  Outliers: {outlier_count}")
    print(f"  Percentage: {outlier_percentage:.2f}%")
    print(f"  Lower Bound: {lower_bound:.2f}")
    print(f"  Upper Bound: {upper_bound:.2f}")
```

```python
df['Is_Male'] = (df['Gender'] == 'Male').astype('uint8')

df = df.drop(columns=['Gender'])

df.head()
```

```python
df.groupby('Performance_Level')['Final_Percentage'].mean()
```

```python
cols_to_drop = ["Student_ID", 'Performance_Level', 'Final_Percentage', 'Math_Score', 'English_Score', 'Science_Score']
df = df.drop(columns=cols_to_drop)
```

```python
df.columns
```

```python
df.info()
```

```python
binary_cols = ['Internet_Access', 'Extracurricular_Activities']
for col in binary_cols:
    df[col] = df[col].astype('bool')
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Pass_Fail'] = le.fit_transform(df['Pass_Fail'])
df['Pass_Fail'] = df['Pass_Fail'].astype('uint8')
```
